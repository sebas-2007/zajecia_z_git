# Zadanie 4

```python
Zrób release 1.0.0.

Na main dodaj literówkę do README.md.

Zrób hotfix readme-typo i popraw błąd.

Zakończ hotfix — zmiana powinna być i w main, i w develop.
```

```python
git flow release start 1.0.0
git commit --allow-empty -m "chore: prepare release 1.0.0"
git flow release finish 1.0.0

echo "Proejct" > README.md
git add README.md && git commit -m "docs: typo"

git flow hotfix start readme-typo
sed -i 's/Proejct/Project/' README.md || perl -0777 -pe 's/Proejct/Project/' -i README.md
git commit -am "fix: correct typo in README"
git flow hotfix finish readme-typo
```

# Zadanie 6

Na develop dopisz Line A do FILE.txt.

W feature f1 dopisz Line B do tego samego pliku.

W feature f2 dopisz Line C do tego samego miejsca.

Spróbuj zakończyć oba feature’y → powstanie konflikt → rozwiąż i dokończ.

```python
echo "Base" > FILE.txt
git add FILE.txt && git commit -m "chore: base"

git flow feature start f1
echo "Line B" >> FILE.txt
git commit -am "feat: add line B"
git flow feature finish f1

git flow feature start f2
echo "Line C" >> FILE.txt
git commit -am "feat: add line C"
git flow feature finish f2   # tu konflikt → rozwiąż plik
git add FILE.txt && git commit --continue
```

# Zadanie 7

Zrób release 2.0.0.

W międzyczasie na main wprowadź błąd.

Zrób hotfix i zakończ go.

Cherry-pick poprawkę również na gałąź release.

```python
git flow release start 2.0.0
git commit --allow-empty -m "chore: prepare release 2.0.0"

git switch main
echo "Buggy" > BUG.txt
git add BUG.txt && git commit -m "feat: buggy code"

git flow hotfix start fix-bug
echo "Fixed" > BUG.txt
git commit -am "fix: remove buggy code"
git flow hotfix finish fix-bug

git switch release/2.0.0
git cherry-pick main
```
